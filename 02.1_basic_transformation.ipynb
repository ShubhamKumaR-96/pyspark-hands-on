```python
from pyspark.sql import SparkSession

spark =  ( SparkSession.builder.appName("Basic Transformation")
          .master("local[*]").getOrCreate())

spark
```

```python
from pyspark.sql.types import (
    StructType, StructField,
    IntegerType, StringType, DateType
)

schema = StructType([
    StructField("employee_id", IntegerType(), True),
    StructField("department_id", IntegerType(), True),
    StructField("name", StringType(), True),
    StructField("age", StringType(), True),
    StructField("gender", StringType(), True),
    StructField("salary", IntegerType(), True),
    StructField("hire_date", DateType(), True)
])

```

```python
emp = spark.read.format("csv")\
      .option("header",True)\
      .schema(schema)\
      .csv("./data/employee.csv")
    
    
```

```python
emp.show()
```

```python
# Column and Expression
from pyspark.sql.functions import col

emp_filtered = emp.select(col("employee_id"),emp["name"],emp.age,emp.salary)

emp_filtered.show()
```

```python
# Using expr for select
from pyspark.sql.functions import expr

emp_casted = emp_filtered.select(expr("employee_id as emp_id"),expr("name as emp_name"),expr("cast(age as int) as age"),emp.salary)
```

```python
emp_casted.show()
```

```python
emp_casted.printSchema()
```

```python
emp_select_expr = emp_casted.selectExpr("emp_id as emp_id" , "emp_name as name","age","salary")
```

```python
emp_select_expr.show()
```

```python
# Filted rmp_based on age > 30

emp_filtered_age = emp_select_expr.select("emp_id","name","age","salary").where("age > 30")
```

```python
emp_filtered_age.show()
```
