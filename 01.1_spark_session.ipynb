```python
from pyspark.sql import SparkSession

spark =  ( SparkSession.builder.appName("Spark Introduction")
          .master("local[*]").getOrCreate())

spark
```

```python
emp_df = (
    spark.read
    .option("header", True)
    .option("inferSchema", True)
    .csv("/home/jupyter/pyspark_hands_on/Interview_Prep/data/employee.csv")
)
```

```python
emp_df.show()
```

```python
emp_final = emp_df.where("salary > 50000")
```

```python
emp_final.rdd.getNumPartitions()
```

```python
emp_final.write.format("noop").mode("overwrite").save()
```

```python
spark.stop()
```
